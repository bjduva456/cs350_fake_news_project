## Download and Extract File

```python
import os
import zipfile
import requests
```

```python
# Define the Kaggle API URL and the destination filename
url = 'https://www.kaggle.com/api/v1/datasets/download/mdepak/fakenewsnet?datasetVersionNumber=1'
zip_file = 'archive.zip'
```

```python
# Download the dataset
response = requests.get(url, stream=True)
with open(zip_file, 'wb') as f:
    f.write(response.content)
```

```python
# Unzip the archive
with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall('fakenewsnet_dataset')
```

```python
# List the extracted files
print(os.listdir('fakenewsnet_dataset'))
```
