## Download and Extract File

```python
import os
import zipfile
import requests
```

```python
# Define the Kaggle API URL and the destination filename
url = 'https://www.kaggle.com/api/v1/datasets/download/mdepak/fakenewsnet?datasetVersionNumber=1'
zip_file = 'archive.zip'
```

```python
# Download the dataset
response = requests.get(url, stream=True)
with open(zip_file, 'wb') as f:
    f.write(response.content)
```

```python
# Unzip the archive
with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall('fakenewsnet_dataset')
```

```python
# List the extracted files
print(os.listdir('fakenewsnet_dataset'))
```

## Missing Values

### buzzfeed_fake_news

```python
buzzfeed_fake_news.isnull().sum()
```

```python
# Fill missing values in 'authors' and 'source' with 'Unknown'
buzzfeed_fake_news['authors'].fillna('Unknown', inplace=True)
buzzfeed_fake_news['source'].fillna('Unknown', inplace=True)
```

```python
# Fill missing values in 'movies' with 0
buzzfeed_fake_news['movies'].fillna(0, inplace=True)
```

```python
# Fill missing values in 'url', 'top_img', 'images', 'canonical_link' with a placeholder like 'No Info'
buzzfeed_fake_news['url'].fillna('No Info', inplace=True)
buzzfeed_fake_news['top_img'].fillna('No Image', inplace=True)
buzzfeed_fake_news['images'].fillna('No Image', inplace=True)
buzzfeed_fake_news['canonical_link'].fillna('No Link', inplace=True)
```

```python
# Fill missing values in 'publish_date' with a placeholder 'Unknown Date'
buzzfeed_fake_news['publish_date'].fillna('Unknown Date', inplace=True)
```

```python
buzzfeed_fake_news.isnull().sum()
```
