## Download and Extract File

```python
import os
import zipfile
import requests
```

```python
# Define the Kaggle API URL and the destination filename
url = 'https://www.kaggle.com/api/v1/datasets/download/mdepak/fakenewsnet?datasetVersionNumber=1'
zip_file = 'archive.zip'
```

```python
# Download the dataset
response = requests.get(url, stream=True)
with open(zip_file, 'wb') as f:
    f.write(response.content)
```

```python
# Unzip the archive
with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall('fakenewsnet_dataset')
```

```python
# List the extracted files
print(os.listdir('fakenewsnet_dataset'))
```

## Visualization

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataframes = [
    buzzfeed_fake_news,
    buzzfeed_real_news,
    politifact_fake_news,
    politifact_real_news
]

# Titles for the heatmaps
titles = [
    "BuzzFeed Fake News",
    "BuzzFeed Real News",
    "PolitiFact Fake News",
    "PolitiFact Real News"
]
```

```python
for i, df in enumerate(dataframes):
    plt.figure(figsize=(12, 6))
    sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
    plt.title(f"Heatmap of Missing Values - {titles[i]}")
    plt.show()
```

```python
buzzfeed_fake_news['source'] = 'BuzzFeed Fake'
buzzfeed_real_news['source'] = 'BuzzFeed Real'
politifact_fake_news['source'] = 'PolitiFact Fake'
politifact_real_news['source'] = 'PolitiFact Real'
```

```python
all_news = pd.concat([buzzfeed_fake_news, buzzfeed_real_news, politifact_fake_news, politifact_real_news])
```

```python
article_counts = all_news['source'].value_counts()
```

```python
plt.figure(figsize=(10, 6))
article_counts.plot(kind='bar', color=['red', 'blue', 'orange', 'green'])
plt.title('Article Counts by Source')
plt.xlabel('Source')
plt.ylabel('Number of Articles')
plt.xticks(rotation=45)
plt.show()
```

```python
def calculate_article_lengths(df):
    return df['text'].apply(len)
```

```python
buzzfeed_fake_lengths = calculate_article_lengths(buzzfeed_fake_news)
buzzfeed_real_lengths = calculate_article_lengths(buzzfeed_real_news)
politifact_fake_lengths = calculate_article_lengths(politifact_fake_news)
politifact_real_lengths = calculate_article_lengths(politifact_real_news)
```

```python
plt.figure(figsize=(14, 8))
```

```python
# BuzzFeed Fake
plt.subplot(2, 2, 1)
plt.hist(buzzfeed_fake_lengths, bins=30, color='red', alpha=0.7)
plt.title('Article Lengths - BuzzFeed Fake')
plt.xlabel('Length (Characters)')
plt.ylabel('Frequency')
```

```python
# BuzzFeed Real
plt.subplot(2, 2, 2)
plt.hist(buzzfeed_real_lengths, bins=30, color='blue', alpha=0.7)
plt.title('Article Lengths - BuzzFeed Real')
plt.xlabel('Length (Characters)')
plt.ylabel('Frequency')
```

```python
# PolitiFact Fake
plt.subplot(2, 2, 3)
plt.hist(politifact_fake_lengths, bins=30, color='orange', alpha=0.7)
plt.title('Article Lengths - PolitiFact Fake')
plt.xlabel('Length (Characters)')
plt.ylabel('Frequency')
```

```python
# PolitiFact Real
plt.subplot(2, 2, 4)
plt.hist(politifact_real_lengths, bins=30, color='green', alpha=0.7)
plt.title('Article Lengths - PolitiFact Real')
plt.xlabel('Length (Characters)')
plt.ylabel('Frequency')
```
